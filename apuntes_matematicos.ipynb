# I. Fundamentos matemáticos formales

## 1. Álgebra lineal avanzada

### Definición 1.1 (Espacio vectorial)

Sea (V) un conjunto no vacío sobre un cuerpo (\mathbb{F}). (V) es un espacio vectorial si cumple los axiomas de clausura, asociatividad, identidad, inverso y distributividad.

**Uso en IA:** los datos, parámetros y representaciones internas de los modelos viven en espacios vectoriales (por ejemplo, embeddings).

---

### Definición 1.2 (Producto interno)

[
\langle x, y \rangle : V \times V \to \mathbb{R}
]

Define una medida de similitud entre vectores.

Induce la norma:
[
|x| = \sqrt{\langle x,x \rangle}
]

**Uso en IA:** base de la similitud coseno, regularización y métricas geométricas.

---

### Teorema 1.1 (Descomposición espectral)

[
A = Q \Lambda Q^T
]

Descompone una matriz simétrica en autovectores y autovalores.

**Uso en IA:** PCA, análisis de estabilidad, reducción de dimensionalidad.

---

### Definición 1.3 (SVD)

[
X = U \Sigma V^T
]

Factoriza cualquier matriz en direcciones principales y escalas.

**Uso en IA:** compresión de datos, embeddings, eliminación de ruido, regularización.

---

## 2. Cálculo multivariado y optimización

### Definición 2.1 (Gradiente)

[
\nabla f = \left( \frac{\partial f}{\partial x_1}, \dots, \frac{\partial f}{\partial x_n} \right)
]

Indica la dirección de máximo incremento de la función.

**Uso en IA:** entrenamiento de modelos mediante descenso por gradiente.

---

### Definición 2.2 (Hessiano)

[
H_f(x) = \left[ \frac{\partial^2 f}{\partial x_i \partial x_j} \right]
]

Describe la curvatura local de la función.

**Uso en IA:** análisis de estabilidad, optimización de segundo orden.

---

### Teorema 2.1 (Condiciones de segundo orden)

Si (\nabla f(x^*)=0) y el Hessiano es definido positivo, entonces (x^*) es un mínimo local.

**Uso en IA:** justifica la convergencia hacia mínimos en entrenamiento.

---

### Ejemplo (Backpropagation)

Capa neuronal:
[
y = \sigma(Wx + b)
]

Gradiente respecto a los pesos:
[
\frac{\partial L}{\partial W} = \frac{\partial L}{\partial y} \cdot \sigma'(z) \cdot x^T
]

**Uso en IA:** permite ajustar los pesos de redes neuronales profundas.

---

## 3. Probabilidad, estadística e inferencia

### Definición 3.1 (Espacio de probabilidad)

((\Omega, \mathcal{F}, P))

Modelo matemático del azar.

**Uso en IA:** base de modelos probabilísticos y generativos.

---

### Teorema 3.1 (Bayes)

[
P(\theta|D) = \frac{P(D|\theta)P(\theta)}{P(D)}
]

Actualiza creencias a partir de datos.

**Uso en IA:** inferencia bayesiana, aprendizaje con incertidumbre.

---

### Definición 3.2 (MLE)

[
\hat{\theta} = \arg\max_\theta \sum_i \log P(x_i|\theta)
]

Busca los parámetros que hacen más probables los datos observados.

**Uso en IA:** entrenamiento clásico de modelos estadísticos.

---

### Definición 3.3 (MAP)

[
\hat{\theta}*{MAP} = \arg\max*\theta \log P(\theta|D)
]

Incorpora conocimiento previo.

**Uso en IA:** regularización bayesiana.

---

## 4. Teoría de la información

### Definición 4.1 (Entropía)

[
H(X) = -\sum_x p(x) \log p(x)
]

Mide incertidumbre.

**Uso en IA:** selección de características, árboles de decisión.

---

### Definición 4.2 (Divergencia KL)

[
D_{KL}(p||q) = \sum p(x) \log \frac{p(x)}{q(x)}
]

Mide diferencia entre distribuciones.

**Uso en IA:** VAEs, regularización, modelos generativos.

# II. Extensión matemática por áreas de IA

## A. Visión por Computador

### Convolución discreta

[
(I * K)(i,j) = \sum_m \sum_n I(i-m,j-n)K(m,n)
]

Aplica un filtro espacial sobre la imagen.

**Uso en IA:** extracción automática de características en CNNs.

---

### Transformada de Fourier

[
F(u,v) = \sum_x \sum_y f(x,y)e^{-2\pi i(ux+vy)}
]

Convierte información espacial en frecuencial.

**Uso en IA:** filtrado, análisis multiescala.

---

## B. Procesamiento de Lenguaje Natural (NLP)

### Modelo probabilístico de lenguaje

[
P(w_1,\dots,w_n) = \prod_i P(w_i|w_{<i})
]

Modela la probabilidad de una secuencia.

**Uso en IA:** base de RNNs, Transformers y LLMs.

---

### Self-Attention

[
\text{Attention}(Q,K,V) = \text{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V
]

Calcula dependencias entre tokens.

**Uso en IA:** núcleo matemático de Transformers.

---

## C. Modelos generativos

### Autoencoder

[
z = f(x), \quad \hat{x} = g(z)
]

Aprende representaciones comprimidas.

**Uso en IA:** reducción de dimensionalidad, detección de anomalías.

---

### Variational Autoencoder (VAE)

ELBO:
[
\mathcal{L} = \mathbb{E}*{q(z|x)}[\log p(x|z)] - D*{KL}(q(z|x)||p(z))
]

Equilibra reconstrucción y regularización.

**Uso en IA:** generación de datos con control probabilístico.

---

### Modelos de difusión

[
q(x_t|x_{t-1}) = \mathcal{N}(\sqrt{1-\beta_t}x_{t-1}, \beta_t I)
]

Agrega ruido progresivamente.

**Uso en IA:** generación de imágenes de alta calidad.

---

## D. Geometric & Graph Deep Learning

### Convolución en grafos

[
H^{(l+1)} = \sigma(\tilde{D}^{-1/2}\tilde{A}\tilde{D}^{-1/2}H^{(l)}W^{(l)})
]

Propaga información entre nodos vecinos.

**Uso en IA:** redes sociales, grafos moleculares, mapas espaciales.

---

## E. IA en dominios físicos

### Physics-Informed Neural Networks (PINNs)

[
L = L_{datos} + \lambda L_{EDP}
]

Combina datos y leyes físicas.

**Uso en IA:** modelado de sistemas físicos, geociencias, astronomía.

---

# III. Ejemplo integrador formal

Clasificador profundo:
[
f_\theta(x) = \text{softmax}(W_2 \sigma(W_1 x))
]

Calcula probabilidades de clase.

Pérdida:
[
L = -\sum y_i \log f_\theta(x_i)
]

Mide error de clasificación.

Actualización:
[
\theta_{t+1} = \theta_t - \eta \nabla L
]

Optimiza los parámetros.
